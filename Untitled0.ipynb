```python
with rasterio.open(file_2026) as src:
    red2026 = src.read(1)
    nir2026 = src.read(2)

print(red2026.shape)
print(nir2026.shape)
```

```python
import rasterio

with rasterio.open(file_2017) as src:
    red2017 = src.read(1)
    nir2017 = src.read(2)

print("Red shape:", red2017.shape)
print("NIR shape:", nir2017.shape)

print("Red minimum:", red2017.min())
print("Red maximum:", red2017.max())

print("NIR minimum:", nir2017.min())
print("NIR maximum:", nir2017.max())
```

```python
import pandas as pd

df = pd.DataFrame({
    "Year": [2017, 2026],
    "Mean_NDVI": [0.3091917487, 0.2979194367]
})

df["Change"] = df["Mean_NDVI"].diff()

df
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 5))

plt.bar(df["Year"], df["Mean_NDVI"])

plt.title("Mean NDVI of Lucknow: 2017 vs 2026")
plt.xlabel("Year")
plt.ylabel("Mean NDVI")

plt.show()
```

```python
import numpy as np
import pandas as pd

# Calculate NDVI
ndvi2017 = (nir2017 - red2017) / (nir2017 + red2017)
ndvi2026 = (nir2026 - red2026) / (nir2026 + red2026)

# Keep only pixels that have valid values in both years
valid = np.isfinite(ndvi2017) & np.isfinite(ndvi2026)

# Get the locations of valid pixels
rows, cols = np.where(valid)

# Pick 1,000 random pixels
np.random.seed(42)

sample_size = min(1000, len(rows))
chosen = np.random.choice(len(rows), sample_size, replace=False)

# Extract their NDVI values
df = pd.DataFrame({
    "Pixel": range(1, sample_size + 1),
    "NDVI_2017": ndvi2017[rows[chosen], cols[chosen]],
    "NDVI_2026": ndvi2026[rows[chosen], cols[chosen]]
})

# Calculate change
df["NDVI_Change"] = df["NDVI_2026"] - df["NDVI_2017"]

print("Dataset size:", len(df))
df.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5))

plt.hist(df["NDVI_2017"], bins=30, alpha=0.6, label="2017")
plt.hist(df["NDVI_2026"], bins=30, alpha=0.6, label="2026")

plt.title("NDVI Distribution: Lucknow 2017 vs 2026")
plt.xlabel("NDVI")
plt.ylabel("Number of Pixels")
plt.legend()

plt.show()
```

```python
# Classify NDVI change

increase = (df["NDVI_Change"] > 0).sum()
decrease = (df["NDVI_Change"] < 0).sum()
no_change = (df["NDVI_Change"] == 0).sum()

categories = ["NDVI Increased", "NDVI Decreased", "No Change"]
counts = [increase, decrease, no_change]

plt.figure(figsize=(8, 5))

plt.bar(categories, counts)

plt.title("Vegetation Change in Sampled Pixels")
plt.xlabel("Change Category")
plt.ylabel("Number of Pixels")

plt.show()

print("Increased:", increase)
print("Decreased:", decrease)
print("No change:", no_change)
```

```python
# Count vegetation changes
increased = (df["NDVI_Change"] > 0).sum()
decreased = (df["NDVI_Change"] < 0).sum()
unchanged = (df["NDVI_Change"] == 0).sum()

total = len(df)

# Convert to percentages
increased_pct = increased / total * 100
decreased_pct = decreased / total * 100
unchanged_pct = unchanged / total * 100

categories = [
    "Vegetation Increased",
    "Vegetation Decreased",
    "Little or No Change"
]

percentages = [
    increased_pct,
    decreased_pct,
    unchanged_pct
]

plt.figure(figsize=(9, 5))

plt.bar(categories, percentages)

plt.title("Change in Vegetation Across 1,000 Sampled Locations")
plt.ylabel("Percentage of Locations (%)")
plt.ylim(0, 100)

plt.show()
```

```python
# Create vegetation categories

def vegetation_level(ndvi):
    if ndvi < 0.2:
        return "Very Low"
    elif ndvi < 0.4:
        return "Low"
    elif ndvi < 0.6:
        return "Moderate"
    else:
        return "High"


df["Vegetation_2017"] = df["NDVI_2017"].apply(vegetation_level)
df["Vegetation_2026"] = df["NDVI_2026"].apply(vegetation_level)


# Count locations in each category

categories = ["Very Low", "Low", "Moderate", "High"]

count_2017 = [
    (df["Vegetation_2017"] == category).sum()
    for category in categories
]

count_2026 = [
    (df["Vegetation_2026"] == category).sum()
    for category in categories
]


# Plot

x = np.arange(len(categories))
width = 0.35

plt.figure(figsize=(9, 5))

plt.bar(x - width/2, count_2017, width, label="2017")
plt.bar(x + width/2, count_2026, width, label="2026")

plt.xticks(x, categories)
plt.xlabel("Vegetation Level")
plt.ylabel("Number of Locations")
plt.title("Vegetation Levels Across 1,000 Sampled Locations")

plt.legend()
plt.show()
```

```python
# Select locations where vegetation decreased
decreased = df[df["NDVI_Change"] < 0].copy()

# Get the 20 locations with the biggest decrease
top_decrease = decreased.sort_values(
    "NDVI_Change"
).head(20).reset_index(drop=True)

top_decrease["Location"] = [
    f"Point {i+1}" for i in range(len(top_decrease))
]

# Plot
plt.figure(figsize=(12, 6))

plt.scatter(
    top_decrease["Location"],
    top_decrease["NDVI_2017"],
    label="2017",
    s=60
)

plt.scatter(
    top_decrease["Location"],
    top_decrease["NDVI_2026"],
    label="2026",
    s=60
)

# Connect 2017 → 2026 for each location
for i in range(len(top_decrease)):
    plt.plot(
        [top_decrease["Location"][i],
         top_decrease["Location"][i]],
        [top_decrease["NDVI_2017"][i],
         top_decrease["NDVI_2026"][i]],
        linestyle="--",
        alpha=0.5
    )

plt.title("Locations with the Largest Vegetation Decrease")
plt.xlabel("Sampled Locations")
plt.ylabel("Vegetation Level")
plt.legend()

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
# Get locations with decreased vegetation
decreased = df[df["NDVI_Change"] < 0].copy()

# Take the 10 biggest decreases
top_decrease = decreased.nsmallest(10, "NDVI_Change")

# Create location labels
top_decrease["Location"] = [
    f"Point {i+1}" for i in range(len(top_decrease))
]

# Plot
x = np.arange(len(top_decrease))
width = 0.35

plt.figure(figsize=(10, 6))

plt.bar(
    x - width/2,
    top_decrease["NDVI_2017"],
    width,
    label="2017"
)

plt.bar(
    x + width/2,
    top_decrease["NDVI_2026"],
    width,
    label="2026"
)

plt.xticks(x, top_decrease["Location"])
plt.xlabel("Location")
plt.ylabel("Vegetation Level")
plt.title("Locations with the Biggest Vegetation Decrease")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Keep only pixels that have valid values in both years
valid = np.isfinite(red2017) & np.isfinite(red2026)

x = red2017[valid]
y = red2026[valid]

# Optional: sample pixels so the graph doesn't become insanely large
sample_size = min(10000, len(x))

np.random.seed(42)
indices = np.random.choice(len(x), sample_size, replace=False)

x_sample = x[indices]
y_sample = y[indices]

plt.figure(figsize=(8, 7))

plt.scatter(
    x_sample,
    y_sample,
    alpha=0.4,
    s=10
)

# Reference line: same Red reflectance in both years
min_val = min(x_sample.min(), y_sample.min())
max_val = max(x_sample.max(), y_sample.max())

plt.plot(
    [min_val, max_val],
    [min_val, max_val],
    linestyle="--"
)

plt.xlabel("Red Band Reflectance — 2017")
plt.ylabel("Red Band Reflectance — 2026")
plt.title("Red Band Comparison: 2017 vs 2026")

plt.grid(alpha=0.2)
plt.show()
```

```python
# Get geographic coordinates from the 2026 TIFF

with rasterio.open(file_2026) as src:
    transform = src.transform

sampled_rows = rows[chosen]
sampled_cols = cols[chosen]

# Convert pixel positions → longitude and latitude
longitudes, latitudes = rasterio.transform.xy(
    transform,
    sampled_rows,
    sampled_cols
)

# Add them to your dataset
df["Latitude"] = latitudes
df["Longitude"] = longitudes

# Check the result
df.head()
```
